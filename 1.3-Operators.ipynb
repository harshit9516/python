Logical Operators

```python
## And, Not , OR
# And operations- whenever both the value of boolean is true so the answer is true and everything is false in AND gate
#And Operations
x=True
y=True

result=x and y
print(result)
```

```python
#OR operations
# whenever both the values you can given it is false and those value is false or false now the other values is true

x=False
y=False

result=x or y
print(result)
```

```python
# NOT Operators
# it is the reverse of the boolean variable the value was given true value is false and false value given is true
X=True
not X
```
